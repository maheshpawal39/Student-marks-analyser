```python
# Convert Order_Date into datetime format
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
```

```python
df.columns
```

```python
# Create useful sales-analysis columns
df["order_year"] = df["order_date"].dt.year
df["order_month"] = df["order_date"].dt.month
df["order_month_name"] = df["order_date"].dt.month_name()
```

```python
df[['order_date','order_year','order_month','order_month_name']].head(10)
```

```python
# Calculate the gross sales before discount
df["gross_sales"] = df["quantity"] * df["unit_price"]
```

```python
df[['quantity','unit_price','gross_sales','discount','sales']].head(10)
```

```python
# Calculate discount amount
df["discount_amount"] = df["gross_sales"] * df["discount"] / 100
```

```python
df[['gross_sales','discount','discount_amount','sales']].head(10)
```

```python
# Check whether calculated net sales matches the Sales column
df["calculated_sales"] = (df["gross_sales"] - df["discount_amount"]).round(2)
(df["calculated_sales"] == df["sales"].round(2)).all()
```

```python
# Create a sales performance category
df["sales_performance"] = pd.qcut(
    df["sales"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
df[['sales','sales_performance']].head(10)
```

```python
# Check final dataset information
df.info()
```

```python
# Save cleaned/transformed dataset
df.to_csv("Sales_Performance_Cleaned.csv", index=False)
```
